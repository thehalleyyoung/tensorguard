`guarded_onnx_export` verifies the module, exports it, and parses the proto back through `onnx.checker` — a structurally invalid graph fails at export time, not at downstream load time.

```python
%pip install -q "git+https://github.com/thehalleyyoung/tensorguard.git"
```

```python
import torch, torch.nn as nn
from tensorguard import guarded_onnx_export

class Net(nn.Module):
    def __init__(self):
        super().__init__()
        self.a = nn.Linear(16, 8)
        self.b = nn.Linear(8, 2)
    def forward(self, x):
        return self.b(torch.relu(self.a(x)))

guarded_onnx_export(Net(), (torch.randn(1, 16),), 'model.onnx',
                    input_shapes={'x': ('batch', 16)})
import onnx; onnx.checker.check_model(onnx.load('model.onnx'))
print('exported + checked OK')
```
