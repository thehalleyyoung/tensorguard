TensorGuard verifies the network inside a `LightningModule`, so training never starts on a shape-broken net.

```python
import torch, torch.nn as nn
import pytorch_lightning as pl
from tensorguard import verify_architecture

class LitMLP(pl.LightningModule):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(20, 10), nn.ReLU(), nn.Linear(10, 2))
    def forward(self, x):
        return self.net(x)

lit = LitMLP()
print('Lightning module ready:', type(lit).__name__)
```

Verify the net's shape contract before `Trainer.fit`:

```python
net_src = '''
import torch, torch.nn as nn
class Net(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(20, 10), nn.ReLU(), nn.Linear(10, 2))
    def forward(self, x):
        return self.net(x)
'''
r = verify_architecture(net_src, input_shapes={'x': ('batch', 20)})
print('Lightning net:', r.status)
assert r.status == 'SAFE'
```
